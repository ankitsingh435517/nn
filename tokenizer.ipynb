```python
text = "Ｕｎｉｃｏｄｅ! 🅤🅝🅘🅒🅞🅓🅔‽ 🇺‌🇳‌🇮‌🇨‌🇴‌🇩‌🇪! 😄 The very name strikes fear and awe into the hearts of programmers worldwide. We all know we ought to “support Unicode” in our software (whatever that means—like using wchar_t for all the strings, right?). But Unicode can be abstruse, and diving into the thousand-page Unicode Standard plus its dozens of supplementary annexes, reports, and notes can be more than a little intimidating. I don’t blame programmers for still finding the whole thing mysterious, even 30 years after Unicode’s inception."
tokens = text.encode("utf-8") # raw bytes
print(tokens)
tokens = list(map(int, tokens)) # convert to a list of integers in range 0..255 for conveninece
print("---")
print(text)
print("Length: ", len(text))
print("---")
print(tokens)
print("Length: ", len(tokens))
```

```python
def get_stats(ids):
  counts = {}
  for pair in zip(ids, ids[1:]):
    counts[pair] = counts.get(pair, 0) + 1
  return counts

stats = get_stats(tokens)
print(stats)
# print(sorted(((v,k) for k,v in stats.items()), reverse=True))
```

```python
top_pair = max(stats, key=stats.get)
print(top_pair)
```

```python
# merge the most frequent pair with a given idx in the list of ids
def merge(ids, pair, idx):
  newids = []
  i = 0
  l = len(ids)
  while i < l:
    if i < l - 1 and ids[i] == pair[0] and ids[i + 1] == pair[1]:
      newids.append(idx)
      i = i + 2
    else:
      newids.append(ids[i])
      i = i + 1

  return newids

merged_ids = merge(tokens, top_pair, len(tokens) + 1)
print(merged_ids)
```

```python
merges = {}
vocab_size = 276
num_ranges = vocab_size - 256

for i in range(num_ranges):
  stats = get_stats(tokens)
  top_pair = max(stats, key=stats.get)
  idx = 256 + i
  merges[top_pair] = idx
  tokens = merge(tokens, top_pair, idx)
```

```python
def encode(text):
  if text is None: return []
  if len(text) == 0: return []
  tokens = list(text.encode("utf-8"))
  while len(tokens) >= 2:
    stats = get_stats(tokens)
    pair = min(stats, key=lambda p: merges.get(p, float("inf")))
    if pair not in merges:
      break
    idx = merges[pair]
    tokens = merge(tokens, pair, idx)
  return tokens

ids = encode("🇺‌")
print(ids)
```

```python
# pre computes and resolve the leaf nodes up (the merges preserve the order of merges as the merges happen iteration wise so for first iteration it merges child nodes at leaf level then next iteration it merges on n-1 level
vocab = {idx: bytes([idx]) for idx in range(256)}
for (p0, p1), idx in merges.items():
  vocab[idx] = vocab[p0] + vocab[p1]

def decode(ids):
  tokens = b"".join(vocab[idx] for idx in ids)
  text = tokens.decode("utf-8", errors="replace")
  return text

decode(encode(text))
```

```python
# adversarial test
# invariant: the invariant here is that for any valid pair of text there has to be encoding in either utf-8 if the merges do not have it encoded byte pair else the unicode for each character in the pair
# and for each valid id there has to be an appropriate python string decoded
assert decode(encode(text)) == text
assert decode([128]) # out of utf-8 schema error handling
assert encode("") == []
assert encode(None) == []
assert decode(encode("01")) == "01"
assert decode(encode("ab")) == "ab"
assert decode(encode("ab")) != "01"
assert decode(encode("aaaaa")) == "aaaaa"
assert decode([]) == ''
assert decode([101]) == 'e'
assert decode([101,32]) == 'e '
assert decode([101,32]) == decode([256])
assert encode("e ") == encode(decode([101,32]))
```
