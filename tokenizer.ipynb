```python
text = "Ｕｎｉｃｏｄｅ! 🅤🅝🅘🅒🅞🅓🅔‽ 🇺‌🇳‌🇮‌🇨‌🇴‌🇩‌🇪! 😄 The very name strikes fear and awe into the hearts of programmers worldwide. We all know we ought to “support Unicode” in our software (whatever that means—like using wchar_t for all the strings, right?). But Unicode can be abstruse, and diving into the thousand-page Unicode Standard plus its dozens of supplementary annexes, reports, and notes can be more than a little intimidating. I don’t blame programmers for still finding the whole thing mysterious, even 30 years after Unicode’s inception."
tokens = text.encode("utf-8") # raw bytes
print(tokens)
tokens = list(map(int, tokens)) # convert to a list of integers in range 0..255 for conveninece
print("---")
print(text)
print("Length: ", len(text))
print("---")
print(tokens)
print("Length: ", len(tokens))
```

```python
def get_stats(ids):
  counts = {}
  for pair in zip(ids, ids[1:]):
    counts[pair] = counts.get(pair, 0) + 1
  return counts

stats = get_stats(tokens)
print(stats)
# print(sorted(((v,k) for k,v in stats.items()), reverse=True))
```

```python
top_pair = max(stats, key=stats.get)
print(top_pair)
```

```python
# merge the most frequent pair with a given idx in the list of ids
def merge(ids, pair, idx):
  newids = []
  i = 0
  l = len(ids)
  while i < l:
    if i < l - 1 and ids[i] == pair[0] and ids[i + 1] == pair[1]:
      newids.append(idx)
      i = i + 2
    else:
      newids.append(ids[i])
      i = i + 1

  return newids

merged_ids = merge(tokens, top_pair, len(tokens) + 1)
print(merged_ids)
```

```python
# repeat the stats -> top_pair -> merge
# repeat until the top_pair fq = 1
stats = get_stats(tokens)
top_pair = max(stats, key=stats.get)
print(top_pair)
max_token = max(tokens)
# print(max)

while stats.get(top_pair) > 3:
  tokens = merge(tokens, top_pair, max_token + 1)
  max_token = max_token + 1
  stats = get_stats(tokens)
  top_pair = max(stats, key=stats.get)

print(tokens)
print(len(tokens))
```

```python
# another way to merge is by merging for certain number of iterations example 20 to reach a desired vocab size say 276 as 256 is already given by unicode
vocab_size = 276 # desired final vocab size
num_merges = vocab_size - 256
ids = list(tokens) # copy of the original tokens to preserve original token ids

merges = {} # (int, int) -> int
for i in range(num_merges):
  stats = get_stats(ids)
  pair = max(stats, key=stats.get)
  idx = 256 + i
  ids = merge(ids, pair, idx)
  merges[pair] = idx # save it for later reverse mapping example (101, 32) -> 256, so for 256 we will get 101, 32 back and those back to their children, so it is building binary child to parent

merges
```

```python
# print(reverse_merges)
# decode: given a list of tokens return a string
def decode_id(id, reverse_merges):
  # e.g 101, check if this is a leaf node if yes then return it in a list [101] else call decode_id on its children
  if id not in reverse_merges:
    return [id]

  c1 = reverse_merges.get(id)[0]
  c2 = reverse_merges.get(id)[1]
  de_ids = [*decode_id(c1, reverse_merges), *decode_id(c2, reverse_merges)]
  return de_ids

def decode(ids, merges={}):
  reverse_merges = {v:k for k,v in merges.items()}
  newids = []
  # we need to recursively call decode for all the ids e.g ids: [101, 119, 256, 312] the recursive function would call decode on each id and would replace 101 with say 32, 91 but then we call decode on 32 and 91 until we reach the leaf as in the base condition is hit where we don't have a parent value in merges
  for id in ids:
    de_ids = decode_id(id, reverse_merges) # this helper function will take a parent id and give its recursively flattened children, then we replace that idx with this
    newids.extend(de_ids)

  return bytes(newids).decode("utf-8")

ids = list(tokens)
s = decode(ids, merges)
print(s)
assert s == text
```

```python
# encode: given a string, encode it to tokens using bpe e.g "Hello World" -> tokens -> compressed token ids (BPE) # preserving merges map for decoding purposes later
def bpe(tokens, num_merges):
  merges = {}
  for i in range(num_merges):
    stats = get_stats(tokens)
    top_pair = max(stats, key=stats.get)
    if stats.get(top_pair) <= 1:
      break
    idx = 256 + i
    tokens = merge(tokens, top_pair, idx)
    merges[top_pair] = idx
  return tokens, merges

def encode(s):
  # first tokenize it as is
  tokens = list(map(int, s.encode("utf-8")))
  # now call bpe to reach 276 or run for 20 times or stop when top_pair frequency is 1
  vocab_size = 276
  num_merges = vocab_size - 256
  if num_merges < 0:
    num_merges = 20
  ids, merges = bpe(tokens, num_merges)
  return ids, merges

ids, merges = encode(text)
print(ids)
s = decode(ids, merges)
print(s)
```
